```python
import re
import pandas as pd


def normalize_year(value):
    if pd.isna(value):
        return None

    value = str(value).strip()

    match = re.search(r"(20\d{2})", value)

    if match:
        return int(match.group(1))

    return None
```

```python
print(normalize_year("FY2024"))
print(normalize_year("2024-25"))
print(normalize_year(2024))
```

```python
def normalize_ticker(value):
    if pd.isna(value):
        return None

    value = str(value).strip().upper()

    value = value.replace("NSE:", "")
    value = value.replace("BSE:", "")

    return value
```

```python
print(normalize_ticker(" reliance "))
print(normalize_ticker("NSE:RELIANCE"))
```
